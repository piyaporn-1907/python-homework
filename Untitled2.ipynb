```python
class Student:
    def __init__(self, student_id, citizen_ID, first_name, last_name, age, gender):
        self.student_id = student_id
        self.citizen_id = citizen_ID
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.gender = gender

    def showInfo (self):
        print("""
        รหัสนักศึกษา:{}
        หมายเลขบัตร 13 หลัก: {}
        ชื่อ: {}
        นามสกุล: {}
        อายุ: {}
        เพศ: {}
        """.format(self.student_id, self.citizen_id, self.first_name, self.last_name, self.age, self.gender))

    def sit(self):
            print("{} is sitting !! ".format(self.first_name))

    def sleep (self):
            print("{} is sleeping !!".format(self.first_name))

    def eat(self):
            print("{} is eatind !!".format(self.first_name))
        

   
```

```python
s1 = Student("6812732133","1103703985795","ปิยาภรณ์", "ทวีสัตย์", 20, "หญิง")
```

```python
s1.showInfo()
s1.sit()
```
