```python
class Dog():
    def _init_(seif, name, gender, gene, age, weight, color):
        self.name = name
        self.gender = gender
        self.gene = gene
        self.age = age
        self.weight = weight
        self.color = color

     def showInfo(self):
         print("""
         ชื่อ: {}
         เพศ: {}
         สายพันธ์: {}
         อายุ: {}
         น้ำหนัก: {}
         สี: {}
         """.format(self.name, self.gender, self.gene, self.age, self.weight, seif.color))
         
         def sit(self):
             print("{} is sitting !!".format(self.name))

         def sleep (self):
             print ("{} is sleeping !! ".format(self.name))

         def eat(self):
             print ("{} is eating !!".format(self.name))
                   
         
```
