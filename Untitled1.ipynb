```python
class Dog(): 
    def __init__(self, name, gender, gene, age, weight, color):
        self.name = name
        self.gender = gender
        self.gene = gene
        self.age = age
        self.weight = weight
        self.color = color
_
    def showInfo(self):
        print("""
ชื่อ: {}
เพศ: {}
สายพันธุ์: {}
อายุ: {}
น้ำหนัก: {}
สี: {}
""".format(self.name, self.gender, self.gene, self.age, self.weight, self.color))

    def sit(self):
        print("{} is sitting !!".format(self.name))

    def sleep(self):
        print("{} is sleeping !!".format(self.name))

    def eat(self):
        print("{} is eating !!".format(self.name))
```

```python
# Instance Variable
dog_goldenswan = Dog ('หงษ์ทอง','famaie', 'โกล์เดน', 4, 80, 'สีขาว')
dog_sangsom = Dog ('แสงโสม', 'maie', 'เซอว์เซอว์', 2, 120, 'สีดำ')
```

```python
dog_sangsom.showInfo()
dog_sangsom.sit()
dog_sangsom.eat()
dog_sangsom.sleep()
```
